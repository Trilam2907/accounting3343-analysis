```python
!pip -q install yfinance ipywidgets lxml html5lib

import os
import re
import time
import math
import pickle
import warnings
from datetime import datetime, timedelta
from concurrent.futures import ThreadPoolExecutor, as_completed

import numpy as np
import pandas as pd
import yfinance as yf
import ipywidgets as widgets
from IPython.display import display, clear_output

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 200)

# Cache files persist in the current Colab runtime. For persistence across
# disconnected runtimes, save them to Google Drive (optional).
CACHE_DIR = "/content/stock_screener_cache"
os.makedirs(CACHE_DIR, exist_ok=True)
CACHE_FILE = os.path.join(CACHE_DIR, "sp500_company_data.pkl")
UNIVERSE_FILE = os.path.join(CACHE_DIR, "sp500_tickers.csv")
CACHE_HOURS = 24

print("Setup complete.")
print("yfinance version:", yf.__version__)
```

```python
def get_sp500_universe(force_refresh=False):
    """Fetch current S&P 500 constituents from Wikipedia and cache the list."""
    if (not force_refresh and os.path.exists(UNIVERSE_FILE)
            and time.time() - os.path.getmtime(UNIVERSE_FILE) < CACHE_HOURS * 3600):
        return pd.read_csv(UNIVERSE_FILE)

    url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
    tables = pd.read_html(url, storage_options={"User-Agent": "Mozilla/5.0 stock-screener-class-project"})
    table = next(t for t in tables if "Symbol" in t.columns and "Security" in t.columns)

    universe = table[["Symbol", "Security", "GICS Sector", "GICS Sub-Industry"]].copy()
    universe.columns = ["Ticker", "Company", "Sector", "Industry"]
    universe["Ticker"] = universe["Ticker"].astype(str).str.replace(".", "-", regex=False)
    universe["Ticker"] = universe["Ticker"].str.strip()
    universe.to_csv(UNIVERSE_FILE, index=False)
    return universe


universe = get_sp500_universe()
print(f"Constituents retrieved: {len(universe)}")
display(universe.head())
```

```python
def _download_one_ticker(ticker):
    """Download one company's Yahoo Finance data; return a serializable record."""
    try:
        t = yf.Ticker(ticker)
        info = t.get_info()  # same underlying information as the .info object
        income = t.financials
        balance = t.balance_sheet
        cashflow = t.cashflow

        return {
            "info": info if isinstance(info, dict) else {},
            "income": income if isinstance(income, pd.DataFrame) else pd.DataFrame(),
            "balance": balance if isinstance(balance, pd.DataFrame) else pd.DataFrame(),
            "cashflow": cashflow if isinstance(cashflow, pd.DataFrame) else pd.DataFrame(),
            "download_error": None
        }
    except Exception as exc:
        return {
            "info": {}, "income": pd.DataFrame(), "balance": pd.DataFrame(),
            "cashflow": pd.DataFrame(), "download_error": str(exc)[:250]
        }


def load_company_data(tickers, force_refresh=False, max_workers=6):
    """Load cached data where possible; download missing/expired tickers."""
    cached = {}
    if not force_refresh and os.path.exists(CACHE_FILE):
        try:
            with open(CACHE_FILE, "rb") as f:
                cached = pickle.load(f)
        except Exception:
            cached = {}

    fresh_enough = {}
    now = time.time()
    if isinstance(cached, dict):
        for ticker, item in cached.items():
            if isinstance(item, dict) and now - item.get("_cached_at", 0) < CACHE_HOURS * 3600:
                fresh_enough[ticker] = item

    results = dict(fresh_enough)
    to_download = [t for t in tickers if t not in results]

    print(f"Using cache: {len([t for t in tickers if t in results])} | To download: {len(to_download)}")
    completed = 0
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {executor.submit(_download_one_ticker, ticker): ticker for ticker in to_download}
        for future in as_completed(futures):
            ticker = futures[future]
            try:
                record = future.result()
            except Exception as exc:
                record = {"info": {}, "income": pd.DataFrame(), "balance": pd.DataFrame(),
                          "cashflow": pd.DataFrame(), "download_error": str(exc)[:250]}
            record["_cached_at"] = time.time()
            results[ticker] = record
            completed += 1
            if completed % 25 == 0 or completed == len(to_download):
                print(f"Downloaded {completed}/{len(to_download)}")

    # Save all current records. Pickle preserves statement DataFrames.
    with open(CACHE_FILE, "wb") as f:
        pickle.dump(results, f)

    return results


# Set FORCE_REFRESH = True to bypass the 24-hour cache and request fresh data.
FORCE_REFRESH = False
company_data = load_company_data(
    universe["Ticker"].tolist(),
    force_refresh=FORCE_REFRESH,
    max_workers=6
)

failed = [t for t, d in company_data.items()
          if d.get("download_error") or not d.get("info")]
print(f"Records in cache: {len(company_data)}")
print(f"Tickers with a download issue or no company info: {len(failed)}")
print("Examples:", failed[:15])
```

```python
def _statement(df):
    """Return statement columns ordered newest to oldest."""
    if not isinstance(df, pd.DataFrame) or df.empty:
        return pd.DataFrame()
    out = df.copy()
    try:
        out = out.reindex(sorted(out.columns, key=lambda x: pd.Timestamp(x), reverse=True), axis=1)
    except Exception:
        pass
    return out


def _get_value(df, aliases, col=0):
    """Find the first available line item alias in a yfinance statement."""
    if df is None or df.empty or df.shape[1] <= col:
        return np.nan
    normalized = {str(idx).strip().lower(): idx for idx in df.index}
    for alias in aliases:
        key = alias.strip().lower()
        if key in normalized:
            value = df.loc[normalized[key]].iloc[col]
            try:
                value = float(value)
                return value if np.isfinite(value) else np.nan
            except (TypeError, ValueError):
                continue
    return np.nan


def _safe_div(numerator, denominator):
    try:
        if pd.isna(numerator) or pd.isna(denominator) or denominator == 0:
            return np.nan
        result = numerator / denominator
        return float(result) if np.isfinite(result) else np.nan
    except Exception:
        return np.nan


def _growth(current, prior):
    """Year-over-year growth; use absolute prior value to avoid sign inversion."""
    if pd.isna(current) or pd.isna(prior) or prior == 0:
        return np.nan
    return _safe_div(current - prior, abs(prior)) * 100


def calculate_company_metrics(ticker, record):
    info = record.get("info", {}) or {}
    inc = _statement(record.get("income"))
    bal = _statement(record.get("balance"))
    cf = _statement(record.get("cashflow"))

    # Income statement: latest annual column (0) and prior annual column (1).
    revenue = _get_value(inc, ["Total Revenue", "Operating Revenue"])
    gross_profit = _get_value(inc, ["Gross Profit"])
    cost_revenue = _get_value(inc, ["Cost Of Revenue", "Cost of Revenue"])
    if pd.isna(gross_profit) and not pd.isna(revenue) and not pd.isna(cost_revenue):
        gross_profit = revenue - cost_revenue

    operating_income = _get_value(inc, ["Operating Income", "Operating Income Loss"])
    ebitda = _get_value(inc, ["EBITDA", "Normalized EBITDA"])
    net_income = _get_value(inc, ["Net Income", "Net Income Common Stockholders",
                                  "Net Income From Continuing Operations"])
    net_income_prior = _get_value(inc, ["Net Income", "Net Income Common Stockholders",
                                        "Net Income From Continuing Operations"], 1)
    eps = _get_value(inc, ["Diluted EPS", "Basic EPS"])
    eps_prior = _get_value(inc, ["Diluted EPS", "Basic EPS"], 1)

    # Balance sheet.
    total_assets = _get_value(bal, ["Total Assets"])
    prior_assets = _get_value(bal, ["Total Assets"], 1)
    equity = _get_value(bal, ["Stockholders Equity", "Total Equity Gross Minority Interest",
                              "Common Stock Equity"])
    prior_equity = _get_value(bal, ["Stockholders Equity", "Total Equity Gross Minority Interest",
                                    "Common Stock Equity"], 1)
    total_debt = _get_value(bal, ["Total Debt"])
    if pd.isna(total_debt):
        current_debt = _get_value(bal, ["Current Debt", "Current Debt And Lease Obligation"])
        long_debt = _get_value(bal, ["Long Term Debt", "Long Term Debt And Capital Lease Obligation"])
        if not pd.isna(current_debt) or not pd.isna(long_debt):
            total_debt = (0 if pd.isna(current_debt) else current_debt) + (0 if pd.isna(long_debt) else long_debt)

    cash = _get_value(bal, ["Cash Cash Equivalents And Short Term Investments",
                            "Cash And Cash Equivalents", "Cash And Short Term Investments"])
    current_assets = _get_value(bal, ["Current Assets"])
    current_liabilities = _get_value(bal, ["Current Liabilities"])

    # Cash flow statement.
    cfo = _get_value(cf, ["Operating Cash Flow", "Cash Flow From Continuing Operating Activities",
                          "Total Cash From Operating Activities"])
    cfo_prior = _get_value(cf, ["Operating Cash Flow", "Cash Flow From Continuing Operating Activities",
                                "Total Cash From Operating Activities"], 1)
    capex = _get_value(cf, ["Capital Expenditure", "Capital Expenditures"])
    capex_prior = _get_value(cf, ["Capital Expenditure", "Capital Expenditures"], 1)

    # Yahoo normally reports capital expenditure as a negative cash flow.
    fcf = cfo + capex if not pd.isna(cfo) and not pd.isna(capex) else np.nan
    fcf_prior = cfo_prior + capex_prior if not pd.isna(cfo_prior) and not pd.isna(capex_prior) else np.nan

    repurchases = _get_value(cf, ["Repurchase Of Capital Stock", "Repurchase Of Stock"])
    issuance = _get_value(cf, ["Common Stock Issuance", "Issuance Of Capital Stock",
                               "Stock Issuance"])
    market_cap = info.get("marketCap", np.nan)

    # Treat cash-flow statement amounts as magnitudes for the net capital-allocation calculation.
    repurchase_amount = abs(repurchases) if not pd.isna(repurchases) else np.nan
    issuance_amount = abs(issuance) if not pd.isna(issuance) else np.nan
    net_repurchase = (repurchase_amount - issuance_amount
                      if not pd.isna(repurchase_amount) and not pd.isna(issuance_amount)
                      else np.nan)
    net_issuance = -net_repurchase if not pd.isna(net_repurchase) else np.nan

    price = info.get("currentPrice", info.get("regularMarketPrice", np.nan))
    trailing_eps = info.get("trailingEps", np.nan)
    pe = info.get("trailingPE", np.nan)
    if pd.isna(pe) and not pd.isna(price) and not pd.isna(trailing_eps) and trailing_eps != 0:
        pe = price / trailing_eps

    avg_assets = ((total_assets + prior_assets) / 2
                  if not pd.isna(total_assets) and not pd.isna(prior_assets) else np.nan)
    avg_equity = ((equity + prior_equity) / 2
                  if not pd.isna(equity) and not pd.isna(prior_equity) else np.nan)

    # STA (scaled total accruals) = (Net Income - Cash Flow from Operations) / Total Assets.
    # This is an accruals-based measure, not the sales-to-assets ratio.
    sta = _safe_div(net_income - cfo, total_assets) if not pd.isna(net_income) and not pd.isna(cfo) else np.nan

    # SNOA proxy = (Total Debt + Total Stockholders' Equity - Cash & Short-Term Investments)
    #             / prior-year Total Assets.
    snoa_numerator = (total_debt + equity - cash
                      if not any(pd.isna(x) for x in [total_debt, equity, cash]) else np.nan)
    snoa = _safe_div(snoa_numerator, prior_assets)

    return {
        "Company": info.get("longName") or info.get("shortName") or ticker,
        "Ticker": ticker,
        "Sector": info.get("sector", np.nan),
        "Industry": info.get("industry", np.nan),
        "Gross Profitability": _safe_div(gross_profit, total_assets),
        "Gross Margin (%)": _safe_div(gross_profit, revenue) * 100,
        "Operating Margin (%)": _safe_div(operating_income, revenue) * 100,
        "EBITDA Margin (%)": _safe_div(ebitda, revenue) * 100,
        "Net Profit Margin (%)": _safe_div(net_income, revenue) * 100,
        "ROA (%)": _safe_div(net_income, avg_assets) * 100,
        "ROE (%)": _safe_div(net_income, avg_equity) * 100,
        "Asset Turnover": _safe_div(revenue, avg_assets),
        "Buyback Yield (%)": _safe_div(net_repurchase, market_cap) * 100,
        "Net Equity Issuance (%)": _safe_div(net_issuance, market_cap) * 100,
        "P/E Ratio": float(pe) if not pd.isna(pe) else np.nan,
        "Debt/Equity": _safe_div(total_debt, equity),
        "Current Ratio": _safe_div(current_assets, current_liabilities),
        "EPS Growth (%)": _growth(eps, eps_prior),
        "FCF Growth (%)": _growth(fcf, fcf_prior),
        "STA (%)": sta * 100 if not pd.isna(sta) else np.nan,
        "SNOA (%)": snoa * 100 if not pd.isna(snoa) else np.nan,
        "Download Status": record.get("download_error") or "OK"
    }


rows = []
for ticker, record in company_data.items():
    rows.append(calculate_company_metrics(ticker, record))

metrics_df = pd.DataFrame(rows)

# Fill missing sector/industry/name with the constituent table where possible.
universe_lookup = universe.set_index("Ticker")
for col in ["Company", "Sector", "Industry"]:
    if col in metrics_df.columns:
        fallback_col = {"Company": "Company", "Sector": "Sector", "Industry": "Industry"}[col]
        metrics_df[col] = metrics_df[col].where(
            metrics_df[col].notna(),
            metrics_df["Ticker"].map(universe_lookup[fallback_col])
        )

# Remove impossible infinities, retain NaN so users can see missing data.
numeric_cols = metrics_df.select_dtypes(include=[np.number]).columns.tolist()
metrics_df[numeric_cols] = metrics_df[numeric_cols].replace([np.inf, -np.inf], np.nan)

print("Companies with calculated rows:", len(metrics_df))
print("Numeric metric coverage:")
display(metrics_df[numeric_cols].notna().sum().sort_values(ascending=False).to_frame("Non-missing values"))
display(metrics_df.head())
```

```python
METRIC_COLUMNS = [
    "Gross Profitability", "Gross Margin (%)", "Operating Margin (%)",
    "EBITDA Margin (%)", "Net Profit Margin (%)", "ROA (%)", "ROE (%)",
    "Asset Turnover", "Buyback Yield (%)", "Net Equity Issuance (%)",
    "P/E Ratio", "Debt/Equity", "Current Ratio", "EPS Growth (%)",
    "FCF Growth (%)", "STA (%)", "SNOA (%)"
]

# Build one range slider per metric, grouped in an Accordion.
metric_sliders = {}
accordion_children = []
group_definitions = {
    "Financial Quality": ["Gross Profitability", "Gross Margin (%)", "Operating Margin (%)",
                          "EBITDA Margin (%)", "Net Profit Margin (%)"],
    "Returns & Efficiency": ["ROA (%)", "ROE (%)", "Asset Turnover"],
    "Shareholder Returns & Capital Allocation": ["Buyback Yield (%)", "Net Equity Issuance (%)"],
    "Valuation": ["P/E Ratio"],
    "Financial Health": ["Debt/Equity", "Current Ratio"],
    "Growth": ["EPS Growth (%)", "FCF Growth (%)"],
    "Additional Quantitative Metrics": ["STA (%)", "SNOA (%)"]
}

def _slider_for_metric(metric):
    series = pd.to_numeric(metrics_df[metric], errors="coerce").dropna()
    if series.empty:
        low, high = 0.0, 1.0
    else:
        low, high = float(series.min()), float(series.max())
        if low == high:
            high = low + max(abs(low) * 0.01, 1.0)
    span = high - low
    step = max(span / 200, 0.0001)
    slider = widgets.FloatRangeSlider(
        value=(low, high), min=low, max=high, step=step,
        description=metric, readout=True, readout_format=".2f",
        continuous_update=False,
        layout=widgets.Layout(width="98%"),
        style={"description_width": "220px"}
    )
    metric_sliders[metric] = slider
    return slider

for group_name, group_metrics in group_definitions.items():
    group_box = widgets.VBox([_slider_for_metric(m) for m in group_metrics])
    accordion_children.append(group_box)

accordion = widgets.Accordion(children=accordion_children)
for i, group_name in enumerate(group_definitions.keys()):
    accordion.set_title(i, group_name)

sector_options = sorted(metrics_df["Sector"].dropna().astype(str).unique().tolist())

sector_select = widgets.SelectMultiple(
    options=sector_options, value=tuple(sector_options),
    description="Sector", rows=7,
    layout=widgets.Layout(width="48%"),
    style={"description_width": "90px"}
)

normalization_select = widgets.Dropdown(
    options=[("No normalized columns", "none"),
             ("Percentile rank (0–100)", "percentile"),
             ("Z-score", "zscore")],
    value="none", description="Comparison:",
    layout=widgets.Layout(width="380px")
)
run_button = widgets.Button(
    description="Run Stock Screen", button_style="success",
    icon="filter", layout=widgets.Layout(width="200px", height="42px")
)
reset_button = widgets.Button(
    description="Reset Filters", button_style="",
    icon="refresh", layout=widgets.Layout(width="150px", height="42px")
)
export_button = widgets.Button(
    description="Export CSV", button_style="info",
    icon="download", disabled=True, layout=widgets.Layout(width="150px", height="42px")
)

results_output = widgets.Output()
screen_state = {"results": pd.DataFrame()}

def run_screen(_=None):
    with results_output:
        clear_output(wait=True)
        result = metrics_df.copy()

        if sector_select.value:
            result = result[result["Sector"].isin(list(sector_select.value))]
        for metric, slider in metric_sliders.items():
            low, high = slider.value
            values = pd.to_numeric(result[metric], errors="coerce")
            result = result[values.notna() & values.between(low, high)]

        mode = normalization_select.value
        if mode == "percentile":
            for metric in METRIC_COLUMNS:
                result[metric + " Percentile"] = metrics_df[metric].rank(pct=True) * 100
                result[metric + " Percentile"] = result[metric + " Percentile"].reindex(result.index)
        elif mode == "zscore":
            for metric in METRIC_COLUMNS:
                mean = metrics_df[metric].mean()
                std = metrics_df[metric].std()
                if pd.isna(std) or std == 0:
                    result[metric + " Z-Score"] = np.nan
                else:
                    result[metric + " Z-Score"] = (result[metric] - mean) / std

        display_cols = ["Company", "Ticker", "Sector", "Industry"] + METRIC_COLUMNS
        if mode == "percentile":
            display_cols += [m + " Percentile" for m in METRIC_COLUMNS]
        elif mode == "zscore":
            display_cols += [m + " Z-Score" for m in METRIC_COLUMNS]

        result = result[display_cols].sort_values("Ticker").reset_index(drop=True)
        screen_state["results"] = result
        export_button.disabled = result.empty

        print(f"Companies matching filters: {len(result):,} of {len(metrics_df):,}")
        if result.empty:
            print("No companies match the current filters. Widen one or more ranges.")
        else:
            try:
                from google.colab import data_table
                display(data_table.DataTable(result, include_index=False, num_rows_per_page=25))
            except Exception:
                display(result)

def reset_filters(_=None):
    for metric, slider in metric_sliders.items():
        slider.value = (slider.min, slider.max)
    sector_select.value = tuple(sector_options)
    normalization_select.value = "none"
    run_screen()

def export_csv(_=None):
    result = screen_state["results"]
    if result.empty:
        print("No results to export.")
        return
    path = "/content/sp500_screen_results.csv"
    result.to_csv(path, index=False)
    try:
        from google.colab import files
        files.download(path)
    except Exception:
        print("CSV saved to:", path)

run_button.on_click(run_screen)
reset_button.on_click(reset_filters)
export_button.on_click(export_csv)

display(widgets.VBox([
    widgets.HTML("<h3>Company universe filters</h3>"),
    widgets.HBox([sector_select]),
    widgets.HTML("<h3>Financial metric ranges</h3>"),
    accordion,
    widgets.HBox([normalization_select]),
    widgets.HBox([run_button, reset_button, export_button]),
    results_output
]))

# Run once with the default broad ranges.
run_screen()
```
