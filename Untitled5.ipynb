```python
import yfinance as yf
import pandas as pd
import numpy as np

def get_value(statement, possible_names, default_value=np.nan):
    """Find a financial statement item using alternative row names."""
    if statement is None or statement.empty:
        return default_value

    for name in possible_names:
        if name in statement.index:
            values = statement.loc[name].dropna()
            if not values.empty:
                return float(values.iloc[0])

    return default_value

results = []

for ticker in tickers:
    try:
        stock = yf.Ticker(ticker)

        income = stock.income_stmt
        cashflow = stock.cashflow
        balance = stock.balance_sheet

        # Net income and operating cash flow
        ni = get_value(income, [
            "Net Income",
            "Net Income Common Stockholders"
        ])

        cfo = get_value(cashflow, [
            "Operating Cash Flow",
            "Cash Flow From Continuing Operating Activities"
        ])

        # Balance sheet data
        assets = get_value(balance, ["Total Assets"])

        cash = get_value(balance, [
            "Cash Cash Equivalents And Short Term Investments",
            "Cash And Cash Equivalents And Short Term Investments",
            "Cash And Cash Equivalents"
        ])

        # Standard debt lines (if missing, we assume 0 instead of failing the formula)
        short_debt = get_value(balance, [
            "Current Debt",
            "Current Debt And Capital Lease Obligation",
            "Current Debt And Finance Lease Obligation",
            "Commercial Paper"
        ], default_value=0.0)

        long_debt = get_value(balance, [
            "Long Term Debt",
            "Long Term Debt And Capital Lease Obligation",
            "Long Term Debt And Finance Lease Obligation"
        ], default_value=0.0)

        # Optional elements: default to 0 if not present on the balance sheet
        minority = get_value(balance, ["Minority Interest", "Non Controlling Interest"], default_value=0.0)
        preferred = get_value(balance, ["Preferred Stock", "Preferred Securities Outside Stockholders Equity"], default_value=0.0)

        common_equity = get_value(balance, [
            "Common Stock Equity",
            "Stockholders Equity",
            "Total Equity Special Stock"
        ])

        # STA: scaled total accruals
        if pd.notna(assets) and assets > 0 and pd.notna(ni) and pd.notna(cfo):
            sta = (ni - cfo) / assets
        else:
            sta = np.nan

        # SNOA: Scaled Net Operating Assets
        if pd.notna(assets) and pd.notna(cash) and pd.notna(common_equity) and assets > 0:
            operating_assets = assets - cash

            operating_liabilities = (
                assets - short_debt - long_debt
                - minority - preferred - common_equity
            )

            snoa = (
                operating_assets - operating_liabilities
            ) / assets
        else:
            snoa = np.nan

        results.append({
            "Ticker": ticker,
            "Net Income": ni,
            "Total Assets": assets,
            "STA": sta,
            "SNOA": snoa
        })

    except Exception as e:
        print(f"Error retrieving {ticker}: {e}")

metrics_df = pd.DataFrame(results)
metrics_df
```

```python
def get_latest_value(statement, possible_names, default_value=np.nan):
    """Return the most recent available value for a statement item."""
    if statement is None or statement.empty:
        return default_value

    for name in possible_names:
        if name in statement.index:
            values = statement.loc[name].dropna()
            if not values.empty:
                return float(values.iloc[0])

    return default_value

altman_results = []

for ticker in tickers:
    try:
        stock = yf.Ticker(ticker)

        balance = stock.balance_sheet
        income = stock.income_stmt
        info = stock.info

        # Balance sheet data
        current_assets = get_latest_value(
            balance, ["Current Assets"]
        )
        current_liabilities = get_latest_value(
            balance, ["Current Liabilities"]
        )
        total_assets = get_latest_value(
            balance, ["Total Assets"]
        )
        retained_earnings = get_latest_value(
            balance, ["Retained Earnings"], default_value=0.0
        )
        total_liabilities = get_latest_value(
            balance, [
                "Total Liabilities Net Minority Interest",
                "Total Liabilities"
            ]
        )

        # Income statement data
        ebit = get_latest_value(
            income, ["EBIT", "Operating Income", "Net Income From Continuing Ops"]
        )
        sales = get_latest_value(
            income, ["Total Revenue", "Interest Income"]
        )

        # Market value of equity
        market_cap = info.get("marketCap", np.nan)

        # Calculate Working Capital (For financial firms like JPM without current assets/liabilities, we can approximate WC as 0 or proxy it)
        if pd.isna(current_assets) or pd.isna(current_liabilities):
            # Financial institutions proxy: Working Capital can be modeled as Cash and cash equivalents minus short term debt if available, or simplified to 0.
            working_capital = 0.0
        else:
            working_capital = current_assets - current_liabilities

        # Inline ternary operators to avoid indentation errors
        x1 = working_capital / total_assets if (pd.notna(working_capital) and pd.notna(total_assets) and total_assets > 0) else 0.0
        x2 = retained_earnings / total_assets if (pd.notna(retained_earnings) and pd.notna(total_assets) and total_assets > 0) else 0.0
        x3 = ebit / total_assets if (pd.notna(ebit) and pd.notna(total_assets) and total_assets > 0) else 0.0
        x4 = market_cap / total_liabilities if (pd.notna(market_cap) and pd.notna(total_liabilities) and total_liabilities > 0) else 0.0
        x5 = sales / total_assets if (pd.notna(sales) and pd.notna(total_assets) and total_assets > 0) else 0.0

        components = [x1, x2, x3, x4, x5]

        if all(pd.notna(x) for x in components):
            z_score = (
                0.012 * (x1 * 100)
                + 0.014 * (x2 * 100)
                + 0.033 * (x3 * 100)
                + 0.006 * (x4 * 100)
                + 0.999 * x5
            )
        else:
            z_score = np.nan

        altman_results.append({
            "Ticker": ticker,
            "Working Capital": working_capital,
            "X1": x1,
            "X2": x2,
            "X3": x3,
            "X4": x4,
            "X5": x5,
            "Altman Z": z_score
        })

    except Exception as e:
        print(f"Error processing {ticker}: {e}")

altman_df = pd.DataFrame(altman_results)
display(altman_df.round(3))
```

```python

def classify_altman_z(z):
    if pd.isna(z):
        return "Insufficient Data"
    elif z > 2.99:
        return "Safe"
    elif z >= 1.81:
        return "Grey"
    else:
        return "Distress"


altman_df["Risk Classification"] = (
    altman_df["Altman Z"].apply(classify_altman_z)
)

display(
    altman_df[
        ["Ticker", "Altman Z", "Risk Classification"]
    ].round(3)
)
```

```python

import yfinance as yf
import pandas as pd
import numpy as np

profitability_results = []

def find_value(statement, names, period):
    """Get a financial statement value for a specific period."""
    if statement is None or statement.empty:
        return np.nan

    for name in names:
        if name in statement.index and period in statement.columns:
            value = statement.loc[name, period]
            if pd.notna(value):
                return float(value)

    return np.nan


for ticker in tickers:
    try:
        stock = yf.Ticker(ticker)

        income = stock.income_stmt
        cashflow = stock.cashflow
        balance = stock.balance_sheet

        if income.empty or cashflow.empty or balance.empty:
            continue

        # Use the two latest periods shared by all three statements
        periods = sorted(
            set(income.columns)
            & set(cashflow.columns)
            & set(balance.columns),
            reverse=True
        )

        if len(periods) < 2:
            continue

        current_period = periods[0]
        previous_period = periods[1]

        # Net income
        ni_current = find_value(
            income,
            ["Net Income", "Net Income Common Stockholders"],
            current_period
        )

        ni_previous = find_value(
            income,
            ["Net Income", "Net Income Common Stockholders"],
            previous_period
        )

        # Operating cash flow
        cfo_current = find_value(
            cashflow,
            ["Operating Cash Flow",
             "Cash Flow From Continuing Operating Activities"],
            current_period
        )

        # Beginning assets for each year's ROA
        assets_current_start = find_value(
            balance, ["Total Assets"], previous_period
        )

        # Assets from the period before the previous year
        assets_previous_start = (
            find_value(balance, ["Total Assets"], periods[2])
            if len(periods) >= 3 else np.nan
        )

        roa_current = (
            ni_current / assets_current_start
            if pd.notna(ni_current)
            and pd.notna(assets_current_start)
            and assets_current_start > 0
            else np.nan
        )

        roa_previous = (
            ni_previous / assets_previous_start
            if pd.notna(ni_previous)
            and pd.notna(assets_previous_start)
            and assets_previous_start > 0
            else np.nan
        )

        cfo_ratio = (
            cfo_current / assets_current_start
            if pd.notna(cfo_current)
            and pd.notna(assets_current_start)
            and assets_current_start > 0
            else np.nan
        )

        # Four binary signals
        score_roa = int(roa_current > 0) if pd.notna(roa_current) else np.nan
        score_cfo = int(cfo_ratio > 0) if pd.notna(cfo_ratio) else np.nan

        delta_roa = (
            roa_current - roa_previous
            if pd.notna(roa_current) and pd.notna(roa_previous)
            else np.nan
        )

        score_delta_roa = (
            int(delta_roa > 0) if pd.notna(delta_roa) else np.nan
        )

        score_accruals = (
            int(cfo_current > ni_current)
            if pd.notna(cfo_current) and pd.notna(ni_current)
            else np.nan
        )

        scores = [
            score_roa, score_cfo,
            score_delta_roa, score_accruals
        ]

        total_score = (
            sum(scores) if all(pd.notna(s) for s in scores)
            else np.nan
        )

        profitability_results.append({
            "Ticker": ticker,
            "ROA": roa_current,
            "CFO / Beginning Assets": cfo_ratio,
            "Change in ROA": delta_roa,
            "ROA Signal": score_roa,
            "CFO Signal": score_cfo,
            "Change ROA Signal": score_delta_roa,
            "Accrual Signal": score_accruals,
            "Profitability Score (0-4)": total_score
        })

    except Exception as e:
        print(f"Could not process {ticker}: {e}")

profitability_df = pd.DataFrame(profitability_results)

display(profitability_df.round(3))
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

funding_results = []

def get_statement_value(statement, names, period, default_value=np.nan):
    if statement is None or statement.empty:
        return default_value

    for name in names:
        if name in statement.index and period in statement.columns:
            value = statement.loc[name, period]
            if pd.notna(value):
                return float(value)

    return default_value

for ticker in tickers:
    try:
        stock = yf.Ticker(ticker)
        balance = stock.balance_sheet
        cashflow = stock.cashflow

        if balance.empty:
            continue

        # Use the latest two shared annual balance-sheet periods
        periods = sorted(balance.columns, reverse=True)

        if len(periods) < 2:
            continue

        current_period = periods[0]
        previous_period = periods[1]

        # Long-term debt and total assets
        debt_now = get_statement_value(
            balance,
            ["Long Term Debt",
             "Long Term Debt And Capital Lease Obligation",
             "Long Term Debt And Finance Lease Obligation"],
            current_period,
            default_value=0.0
        )

        debt_before = get_statement_value(
            balance,
            ["Long Term Debt",
             "Long Term Debt And Capital Lease Obligation",
             "Long Term Debt And Finance Lease Obligation"],
            previous_period,
            default_value=0.0
        )

        assets_now = get_statement_value(
            balance, ["Total Assets"], current_period
        )

        assets_before = get_statement_value(
            balance, ["Total Assets"], previous_period
        )

        leverage_now = (
            debt_now / assets_now
            if pd.notna(debt_now) and pd.notna(assets_now) and assets_now > 0
            else 0.0
        )

        leverage_before = (
            debt_before / assets_before
            if pd.notna(debt_before) and pd.notna(assets_before) and assets_before > 0
            else 0.0
        )

        leverage_change = (
            leverage_now - leverage_before
            if pd.notna(leverage_now) and pd.notna(leverage_before)
            else 0.0
        )

        leverage_score = (
            int(leverage_change < 0)
            if pd.notna(leverage_change) else 0
        )

        # Current ratio (handling financial institutions which do not report current items)
        ca_now = get_statement_value(
            balance, ["Current Assets"], current_period
        )
        cl_now = get_statement_value(
            balance, ["Current Liabilities"], current_period
        )
        ca_before = get_statement_value(
            balance, ["Current Assets"], previous_period
        )
        cl_before = get_statement_value(
            balance, ["Current Liabilities"], previous_period
        )

        # If bank, proxy current assets with cash + near-cash items and current liabilities with short term liabilities/deposits
        if pd.isna(ca_now) or pd.isna(cl_now):
            cash_now = get_statement_value(balance, ["Cash And Cash Equivalents", "Cash Cash Equivalents And Short Term Investments"], current_period, 0.0)
            short_liab_now = get_statement_value(balance, ["Current Debt", "Current Liabilities"], current_period, 1.0)
            current_ratio_now = cash_now / short_liab_now if short_liab_now > 0 else 1.0
        else:
            current_ratio_now = ca_now / cl_now if cl_now > 0 else 1.0

        if pd.isna(ca_before) or pd.isna(cl_before):
            cash_before = get_statement_value(balance, ["Cash And Cash Equivalents", "Cash Cash Equivalents And Short Term Investments"], previous_period, 0.0)
            short_liab_before = get_statement_value(balance, ["Current Debt", "Current Liabilities"], previous_period, 1.0)
            current_ratio_before = cash_before / short_liab_before if short_liab_before > 0 else 1.0
        else:
            current_ratio_before = ca_before / cl_before if cl_before > 0 else 1.0

        liquidity_change = (
            current_ratio_now - current_ratio_before
            if pd.notna(current_ratio_now) and pd.notna(current_ratio_before)
            else 0.0
        )

        liquidity_score = (
            int(liquidity_change > 0)
            if pd.notna(liquidity_change) else 0
        )

        # Equity issuance proxy: financing cash flow check
        equity_issuance = get_statement_value(cashflow, ["Proceeds From Stock Option Exercises", "Net Common Stock Issuance"], current_period, 0.0)
        equity_offer_score = int(equity_issuance <= 0)
        equity_offer_status = "No Issuance" if equity_offer_score == 1 else "Issued Equity"

        funding_results.append({
            "Ticker": ticker,
            "Leverage (Current)": leverage_now,
            "Leverage (Previous)": leverage_before,
            "Leverage Change": leverage_change,
            "Leverage Score": leverage_score,
            "Current Ratio (Current)": current_ratio_now,
            "Current Ratio (Previous)": current_ratio_before,
            "Liquidity Change": liquidity_change,
            "Liquidity Score": liquidity_score,
            "Equity Issuance Score": equity_offer_score,
            "Equity Issuance Status": equity_offer_status
        })

    except Exception as e:
        print(f"Could not process {ticker}: {e}")

funding_df = pd.DataFrame(funding_results)
display(funding_df.round(3))
```

```python
efficiency_results = []

for ticker in tickers:
    try:
        stock = yf.Ticker(ticker)
        income = stock.income_stmt
        balance = stock.balance_sheet

        if income.empty or balance.empty:
            continue

        income_periods = sorted(income.columns, reverse=True)
        balance_periods = sorted(balance.columns, reverse=True)

        if len(income_periods) < 2 or len(balance_periods) < 3:
            continue

        current_year = income_periods[0]
        previous_year = income_periods[1]

        # Sales for both years
        sales_current = get_statement_value(
            income, ["Total Revenue"], current_year
        )
        sales_previous = get_statement_value(
            income, ["Total Revenue"], previous_year
        )

        # Gross profit (with Total Revenue as a fallback proxy if Gross Profit index doesn't exist, common for banks)
        gross_current = get_statement_value(
            income, ["Gross Profit"], current_year,
            default_value=sales_current
        )
        gross_previous = get_statement_value(
            income, ["Gross Profit"], previous_year,
            default_value=sales_previous
        )

        # Beginning assets for each fiscal year
        assets_begin_current = get_statement_value(
            balance, ["Total Assets"], balance_periods[1]
        )
        assets_begin_previous = get_statement_value(
            balance, ["Total Assets"], balance_periods[2]
        )

        # Gross margin change
        margin_change = np.nan
        if (pd.notna(gross_current) and
            pd.notna(gross_previous) and
            pd.notna(sales_current) and sales_current > 0 and
            pd.notna(sales_previous) and sales_previous > 0):

            margin_change = (
                gross_current / sales_current
                - gross_previous / sales_previous
            )
        else:
            margin_change = 0.0

        # Asset turnover change
        turnover_change = np.nan
        if (pd.notna(sales_current) and
            pd.notna(sales_previous) and
            pd.notna(assets_begin_current) and
            pd.notna(assets_begin_previous) and
            assets_begin_current > 0 and assets_begin_previous > 0):

            turnover_change = (
                sales_current / assets_begin_current
                - sales_previous / assets_begin_previous
            )
        else:
            turnover_change = 0.0

        margin_score = (
            int(margin_change > 0)
            if pd.notna(margin_change) else 0
        )

        turnover_score = (
            int(turnover_change > 0)
            if pd.notna(turnover_change) else 0
        )

        efficiency_score = (
            margin_score + turnover_score
            if pd.notna(margin_score) and pd.notna(turnover_score)
            else 0
        )

        efficiency_results.append({
            "Ticker": ticker,
            "Change in Gross Margin": margin_change,
            "Margin Indicator": margin_score,
            "Change in Asset Turnover": turnover_change,
            "Turnover Indicator": turnover_score,
            "Efficiency Score (0-2)": efficiency_score
        })

    except Exception as e:
        print(f"Could not process {ticker}: {e}")

efficiency_df = pd.DataFrame(efficiency_results)

display(efficiency_df.round(4))
```

```python

# Merge the three financial-strength score tables
master_df = (
    profitability_df[
        ["Ticker", "Profitability Score (0-4)"]
    ]
    .merge(
        funding_df[
            ["Ticker", "Leverage Score",
             "Liquidity Score", "Equity Issuance Score"]
        ],
        on="Ticker",
        how="outer"
    )
    .merge(
        efficiency_df[
            ["Ticker", "Efficiency Score (0-2)"]
        ],
        on="Ticker",
        how="outer"
    )
)

score_columns = [
    "Profitability Score (0-4)",
    "Leverage Score",
    "Liquidity Score",
    "Equity Issuance Score",
    "Efficiency Score (0-2)"
]

# Only calculate a total when every component is available
master_df["Financial Strength Score (0-9)"] = (
    master_df[score_columns].sum(axis=1, min_count=len(score_columns))
)

display(
    master_df.sort_values(
        "Financial Strength Score (0-9)",
        ascending=True,
        na_position="last"
    )
)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

def get_value(statement, row, period):
    """Return a financial statement value, or NaN if unavailable."""
    if statement is None or statement.empty:
        return np.nan
    if row not in statement.index or period not in statement.columns:
        return np.nan

    value = statement.loc[row, period]

    if isinstance(value, pd.Series):
        value = value.iloc[0]

    if pd.isna(value):
        return np.nan

    return float(value)


def first_available(statement, rows, period):
    """Try alternative row names when a field is unavailable."""
    for row in rows:
        value = get_value(statement, row, period)
        if pd.notna(value):
            return value
    return np.nan


def calculate_metrics(ticker):
    stock = yf.Ticker(ticker)

    income = stock.financials
    balance = stock.balance_sheet
    cashflow = stock.cashflow

    # Check data availability
    if income.empty or balance.empty or cashflow.empty:
        return {"Ticker": ticker, "Error": "Missing financial statements"}

    # Use the three most recent common reporting dates
    periods = sorted(
        set(income.columns) & set(balance.columns) & set(cashflow.columns),
        reverse=True
    )

    if len(periods) < 2:
        return {"Ticker": ticker, "Error": "Insufficient common periods"}

    current = periods[0]
    previous = periods[1]
    two_years_ago = periods[2] if len(periods) >= 3 else previous

    # Helper to retrieve values from the three statements
    def inc(row, period):
        return get_value(income, row, period)

    def bal(row, period):
        return get_value(balance, row, period)

    def cf(row, period):
        return get_value(cashflow, row, period)

    # Basic financial values
    ni = first_available(
        income,
        ["Net Income", "Net Income Common Stockholders", "Net Income Continuous Operations"],
        current
    )
    if pd.isna(ni): ni = 0.0

    ni_prev = first_available(
        income,
        ["Net Income", "Net Income Common Stockholders", "Net Income Continuous Operations"],
        previous
    )
    if pd.isna(ni_prev): ni_prev = 0.0

    cfo = first_available(
        cashflow,
        ["Operating Cash Flow", "Cash Flow From Continuing Operating Activities"],
        current
    )
    if pd.isna(cfo): cfo = 0.0

    assets = bal("Total Assets", current)
    assets_prev = bal("Total Assets", previous)
    assets_2prev = bal("Total Assets", two_years_ago)
    if pd.isna(assets_prev): assets_prev = assets
    if pd.isna(assets_2prev): assets_2prev = assets_prev

    sales = first_available(
        income, ["Total Revenue", "Operating Revenue"], current
    )
    if pd.isna(sales): sales = 0.0

    sales_prev = first_available(
        income, ["Total Revenue", "Operating Revenue"], previous
    )
    if pd.isna(sales_prev): sales_prev = sales

    gross_profit = first_available(
        income, ["Gross Profit"], current
    )
    if pd.isna(gross_profit):
        gross_profit = sales  # Proxy for banks / financial services

    gross_profit_prev = first_available(
        income, ["Gross Profit"], previous
    )
    if pd.isna(gross_profit_prev):
        gross_profit_prev = sales_prev

    ebit = first_available(
        income, ["EBIT", "Operating Income", "Total Operating Income As Reported"], current
    )
    if pd.isna(ebit):
        ebit = ni # fallback to net income

    # STA: accruals scaled by total assets
    sta = (
        (ni - cfo) / assets
        if pd.notna(ni) and pd.notna(cfo) and pd.notna(assets) and assets != 0
        else 0.0
    )

    # SNOA: using the operating asset/liability approximation
    cash = first_available(
        balance,
        ["Cash Cash Equivalents And Short Term Investments",
         "Cash And Cash Equivalents", "Cash And Cash Equivalents And Short Term Investments"],
        current
    )
    if pd.isna(cash): cash = 0.0

    current_debt = first_available(
        balance,
        ["Current Debt And Capital Lease Obligation", "Current Debt", "Current Debt And Finance Lease Obligation", "Commercial Paper"],
        current
    )
    if pd.isna(current_debt): current_debt = 0.0

    long_debt = first_available(
        balance,
        ["Long Term Debt And Capital Lease Obligation", "Long Term Debt", "Long Term Debt And Finance Lease Obligation"],
        current
    )
    if pd.isna(long_debt): long_debt = 0.0

    minority = first_available(balance, ["Minority Interest", "Non Controlling Interest"], current)
    if pd.isna(minority): minority = 0.0

    equity = first_available(
        balance, ["Stockholders Equity", "Common Stock Equity", "Total Equity Gross Minority Interest", "Total Equity Special Stock"],
        current
    )
    if pd.isna(equity): equity = assets - long_debt - current_debt

    snoa = 0.0
    if pd.notna(assets) and assets != 0 and pd.notna(equity):
        operating_assets = assets - cash
        operating_liabilities = (
            assets - current_debt - long_debt - minority - equity
        )
        snoa = (operating_assets - operating_liabilities) / assets

    # Altman Z-score with intelligent fallbacks
    working_capital = bal("Working Capital", current)
    if pd.isna(working_capital):
        ca = bal("Current Assets", current)
        cl = bal("Current Liabilities", current)
        if pd.notna(ca) and pd.notna(cl):
            working_capital = ca - cl
        else:
            working_capital = 0.0 # Standard bank/financial proxy

    retained = first_available(balance, ["Retained Earnings"], current)
    if pd.isna(retained): retained = 0.0

    liabilities = first_available(balance, ["Total Liabilities Net Minority Interest", "Total Liabilities"], current)
    if pd.isna(liabilities): liabilities = assets - equity if pd.notna(equity) else assets

    try:
        market_cap = stock.info.get("marketCap", np.nan)
    except Exception:
        market_cap = np.nan
    if pd.isna(market_cap):
        market_cap = assets

    z = 0.0
    if pd.notna(assets) and assets > 0:
        x1 = working_capital / assets
        x2 = retained / assets
        x3 = ebit / assets
        x4 = market_cap / liabilities if pd.notna(liabilities) and liabilities > 0 else 1.0
        x5 = sales / assets if pd.notna(sales) else 0.0

        z = 1.2 * x1 + 1.4 * x2 + 3.3 * x3 + 0.6 * x4 + 1.0 * x5

    if z > 2.99:
        z_class = "Safe"
    elif z >= 1.81:
        z_class = "Grey"
    else:
        z_class = "Distress"

    # Profitability signals (4)
    roa = ni / assets_prev if (pd.notna(ni) and pd.notna(assets_prev) and assets_prev != 0) else 0.0
    cfo_roa = cfo / assets_prev if (pd.notna(cfo) and pd.notna(assets_prev) and assets_prev != 0) else 0.0
    roa_prev = ni_prev / assets_2prev if (pd.notna(ni_prev) and pd.notna(assets_2prev) and assets_2prev != 0) else 0.0

    p1 = int(roa > 0)
    p2 = int(cfo_roa > 0)
    p3 = int(roa > roa_prev)
    p4 = int(cfo > ni) if pd.notna(cfo) and pd.notna(ni) else 0

    profitability_score = sum([p1, p2, p3, p4])

    # Leverage, liquidity, and funding signals (3)
    long_debt_prev = first_available(
        balance,
        ["Long Term Debt And Capital Lease Obligation", "Long Term Debt", "Long Term Debt And Finance Lease Obligation"],
        previous
    )
    if pd.isna(long_debt_prev): long_debt_prev = 0.0

    leverage = long_debt / assets if (pd.notna(assets) and assets != 0) else 0.0
    leverage_prev = long_debt_prev / assets_prev if (pd.notna(assets_prev) and assets_prev != 0) else 0.0

    current_ratio = (
        bal("Current Assets", current) / bal("Current Liabilities", current)
        if pd.notna(bal("Current Assets", current)) and pd.notna(bal("Current Liabilities", current)) and bal("Current Liabilities", current) != 0
        else 1.0
    )
    current_ratio_prev = (
        bal("Current Assets", previous) / bal("Current Liabilities", previous)
        if pd.notna(bal("Current Assets", previous)) and pd.notna(bal("Current Liabilities", previous)) and bal("Current Liabilities", previous) != 0
        else 1.0
    )

    l1 = int(leverage < leverage_prev)
    l2 = int(current_ratio > current_ratio_prev)

    net_issuance = first_available(cashflow, ["Net Common Stock Issuance", "Proceeds From Stock Option Exercises"], current)
    l3 = int(net_issuance <= 0) if pd.notna(net_issuance) else 1

    funding_score = sum([l1, l2, l3])

    # Operating efficiency signals (2)
    margin = gross_profit / sales if (pd.notna(sales) and sales != 0) else 0.0
    margin_prev = gross_profit_prev / sales_prev if (pd.notna(sales_prev) and sales_prev != 0) else 0.0

    turnover = sales / assets_prev if (pd.notna(assets_prev) and assets_prev != 0) else 0.0
    turnover_prev = sales_prev / assets_2prev if (pd.notna(assets_2prev) and assets_2prev != 0) else 0.0

    e1 = int(margin > margin_prev)
    e2 = int(turnover > turnover_prev)
    efficiency_score = sum([e1, e2])

    total_score = profitability_score + funding_score + efficiency_score

    return {
        "Ticker": ticker,
        "Latest Fiscal Year": current.year if hasattr(current, 'year') else current,
        "STA": sta,
        "SNOA": snoa,
        "Altman Z": z,
        "Risk Classification": z_class,
        "Profitability Score (0-4)": profitability_score,
        "Leverage Score": l1,
        "Liquidity Score": l2,
        "Equity Issuance Score": l3,
        "Funding Score (0-3)": funding_score,
        "Efficiency Score (0-2)": efficiency_score,
        "Financial Strength Score (0-9)": total_score
    }
```

```python

test_tickers = ["AAPL", "MSFT", "NVDA", "KO", "JPM"]

test_results = []

for ticker in test_tickers:
    try:
        test_results.append(calculate_metrics(ticker))
    except Exception as e:
        test_results.append({
            "Ticker": ticker,
            "Error": str(e)
        })

test_df = pd.DataFrame(test_results)

display(test_df.round(3))
```

```python

import time
import pandas as pd

results = []
errors = []

for i, ticker in enumerate(tickers, start=1):
    try:
        result = calculate_metrics(ticker)
        results.append(result)

    except Exception as e:
        errors.append({
            "Ticker": ticker,
            "Error": str(e)
        })

    # Show progress every 25 companies
    if i % 25 == 0:
        print(f"Processed {i}/{len(tickers)} companies")

    time.sleep(0.1)  # Be gentle with data requests

# Create the full results table
screen_df = pd.DataFrame(results)

# Save any errors for troubleshooting
errors_df = pd.DataFrame(errors)

print("\nCompanies processed:", len(results))
print("Errors:", len(errors))

display(screen_df.head(10))
```

```python

# Keep companies with an available financial strength score
ranked_df = screen_df.dropna(
    subset=["Financial Strength Score (0-9)"]
).copy()

# Lower scores indicate weaker performance on these nine signals
ranked_df = ranked_df.sort_values(
    by="Financial Strength Score (0-9)",
    ascending=True
)

columns_to_show = [
    "Ticker",
    "STA",
    "SNOA",
    "Altman Z",
    "Risk Classification",
    "Financial Strength Score (0-9)"
]

display(ranked_df[columns_to_show].head(20).round(3))
```
